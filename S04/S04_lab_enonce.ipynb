## Partie 2 – Automates à états (45 min)

> **Rappel**
> - Un automate : des **états**, un état initial, des **transitions** déclenchées par des **événements**
> - En Python : `state = "RED"` + un dictionnaire de transitions `transitions[state] -> état suivant`
> - Reconnaissance de chaîne : on lit caractère par caractère ; un état `ERROR` absorbe tout ; on **accepte** si l'état final est un état final valide

### Exemple guidé : workflow d'une demande d'achat

```python
transitions = {
    "DRAFT":     {"submit": "SUBMITTED"},
    "SUBMITTED": {"approve": "APPROVED", "reject": "DRAFT"},
    "APPROVED":  {"order": "ORDERED"},
    "ORDERED":   {},
}
state = "DRAFT"
for event in ["submit", "reject", "submit", "order", "approve", "order"]:
    allowed = transitions[state]
    if event in allowed:
        print(f"{state:<9} --{event}--> {allowed[event]}")
        state = allowed[event]
    else:
        print(f"{state:<9} : événement '{event}' refusé")
print("État final :", state)
```
