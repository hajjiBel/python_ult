## Partie 2 – Séquences, dictionnaires, ensembles et références

> **Rappel**
> - Indexation `s[0]`, `s[-1]` · slicing `s[début:fin:pas]` (fin exclue)
> - Listes : `append`, `insert`, `remove`, `pop`, `sort`, `len` · chaînes : `split`, `strip`, `upper`, `join`
> - `dict` : `d["clé"]`, `d.get("clé", défaut)`, `d.keys()`, `d.values()`, `d.items()`
> - `set` : `&` intersection, `|` union, `-` différence
> - `b = a` crée un **alias** ; `a.copy()` copie superficielle ; `copy.deepcopy(a)` copie profonde
> - Unpacking : `a, b, c = seq` · `first, *rest = seq`

### Exemple guidé : analyser une ligne de journal

````python
line = "2026-10-19 23:05:12 ERROR srv-db-01 disk full on /dev/sda1"
date = line[:10]
time = line[11:19]
words = line.split()
level, host = words[2], words[3]
message = " ".join(words[4:])
print(date, time, level, host, "->", message)
print(len(words), "mots ; message en majuscules :", message.upper())
````
