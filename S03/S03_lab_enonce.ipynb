## Partie 2 – Boucles

> **Rappel**
> - `while cond:` répète tant que la condition est vraie · `x += 1`
> - `for item in seq:` · `range(début, fin, pas)` · `enumerate(seq, start=1)` · `zip(a, b)`
> - `for k, v in d.items():`
> - `break` sort de la boucle, `continue` passe au tour suivant, `pass` ne fait rien
> - `for ... else:` le `else` s'exécute si aucun `break`

### Exemple guidé : inventaire par catégorie

````python
inventory = [
    {"asset_tag": "LT-001", "category": "laptop", "price": 3200.0},
    {"asset_tag": "LT-002", "category": "laptop", "price": 2900.0},
    {"asset_tag": "SV-001", "category": "server", "price": 18500.0},
]
counts = {}
total = 0.0
for item in inventory:
    category = item["category"]
    counts[category] = counts.get(category, 0) + 1
    total += item["price"]

for category, count in counts.items():
    print(f"{category:<8} {count}")
print(f"Valeur totale : {total:,.2f} DT")
````
